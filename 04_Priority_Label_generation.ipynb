##Priority Label Generation

####Business Rule for Priority Assignment

The original dataset contains customer support tickets with four attributes: instruction, response, category, and intent. However, it does not include a priority field, which is an important part of a customer support system.

To address this limitation, a new column named priority is generated using rule-based business logic. Instead of assigning priorities randomly, each ticket category is mapped to a priority level based on the urgency, business impact, and expected response time of the issue.

The priority levels are defined as follows:

High Priority: Tickets related to financial transactions or account access, such as Payment, Refund, and Account issues or a direct request to talk to customer support agent. These requests require immediate attention because they directly affect customer finances or access to services.


Medium Priority: Tickets related to operational processes, including Order, Shipping, Delivery, and Cancellation. These issues are important and time-sensitive but generally do not involve immediate financial or security risks.


Low Priority: General inquiries and informational requests, such as Invoice, Feedback, and Subscription. These requests typically do not require immediate action and have a comparatively lower business impact.

This rule-based labeling provides a consistent and explainable target variable for training the priority prediction model.

```python
priority_map = {
    # High
    "complaint": "High",
    "payment_issue": "High",
    "contact_human_agent": "High",
    "contact_customer_service": "High",
    "check_refund_policy": "High",
    "cancel_order": "High",
    "check_cancellation_fee": "High",
    "registration_problems": "High",

    # Medium
    "place_order": "Medium",
    "change_order": "Medium",
    "track_order": "Medium",
    "delivery_period": "Medium",
    "delivery_options": "Medium",
    "create_account": "Medium",
    "recover_password": "Medium",
    "set_up_shipping_address": "Medium",
    "change_shipping_address": "Medium",
    "track_refund": "Medium",
    "get_refund": "Medium",

    # Low
    "delete_account": "Low",
    "switch_account": "Low",
    "edit_account": "Low",
    "check_invoice": "Low",
    "get_invoice": "Low",
    "check_payment_methods": "Low",
    "review": "Low",
    "newsletter_subscription": "Low"
}

df["priority"] = df["intent"].map(priority_map)
```

```python
df["priority"] = df["intent"].map(priority_map)
```
