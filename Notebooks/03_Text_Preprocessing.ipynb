All the preprocessing will be done in the instruction and response column only. So in this we will refer df as these two columns only

```python
import numpy as np
import pandas as pd
from google.colab import files
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
df = pd.read_csv('/content/Bitext_Sample_Customer_Support_Training_Dataset.csv')
df = df[['instruction', 'response']]
df
```

```python
df = df.dropna()
df
```

# 1. Lowering Text

```python
df = df.apply(lambda x : x.str.lower())
```

# 2. Remove Punctuations

```python
import re

df["instruction"] = df["instruction"].apply(
    lambda text: re.sub(r"[^\w\s]", "", text)
)
df
```

# 3. Remove whitespaces

```python
df = df.apply(lambda x : x.str.strip())
df
```

# 4. Tokenization

```python
from nltk.tokenize import word_tokenize
df=df.apply(lambda x:x.apply(word_tokenize))
df
```

```python
stop_words = set(stopwords.words('english'))

df = df.apply(
    lambda col: col.apply(
        lambda words: [word for word in words if word.lower() not in stop_words]
    )
)
df
```

# 5. Lemmatization

```python
lemmatizer = WordNetLemmatizer()

df = df.apply(
    lambda col: col.apply(
        lambda words: [
            lemmatizer.lemmatize(word, pos=wordnet.VERB)
            for word in words
        ]
    )
)
df
```

# 6. Spelling Correction

```python
!pip install pyspellchecker
from spellchecker import SpellChecker
spell = SpellChecker()

# Build unique word set directly from token lists
unique_words = set()
for col in ["instruction", "response"]:
    for tokens in df[col]:
        unique_words.update(tokens)

# Correct each unique word once
correction_cache = {w: (spell.correction(w) or w) for w in unique_words}

# Apply correction to each token in the list, return a corrected list (not a joined string)
def correct_tokens(tokens):
    return [correction_cache[w] for w in tokens]

df["instruction"] = df["instruction"].apply(correct_tokens)
df["response"] = df["response"].apply(correct_tokens)
```
